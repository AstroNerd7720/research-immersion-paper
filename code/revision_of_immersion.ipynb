```python
df_filtered = df.copy()

# Require a valid planetary radius and both uncertainty values
valid_radius = (
    df_filtered['pl_rade'].notna() &
    (df_filtered['pl_rade'] > 0) &
    df_filtered[['pl_radeerr1', 'pl_radeerr2']].notna().all(axis=1)
)

df_filtered = df_filtered[valid_radius].copy()

# Use the larger side of the asymmetric uncertainty
df_filtered['rad_frac_err'] = (
    df_filtered[['pl_radeerr1', 'pl_radeerr2']]
    .abs()
    .max(axis=1)
    / df_filtered['pl_rade']
)

# Keep planets with <= 20% fractional radius uncertainty
df_filtered = df_filtered[
    df_filtered['rad_frac_err'] <= 0.20
].copy()

print(f'Total Sample Size after Filtering: {len(df_filtered)}')

df_filtered.to_csv('nea_filtered_2026-10-03.csv', index=False)
```

The see if planetary types has a relationship with the star it orbits

```python
contingencytable = pd.crosstab(df_stack['Star Classification'], df_stack['Planet Classification'])


ax = contingencytable.plot(kind='bar', stacked=True, figsize=(10, 6), edgecolor='black', color=["#66c2a5", "#e78ac3","#8da0cb" , "#fc8d62", "#ffd92f" ])


plt.title('Stacked Distribution of Planets by Star Type', fontsize=12, fontweight='bold')
plt.xlabel('Star Type', fontsize=12)
plt.xticks(rotation=0)
plt.ylabel('Number of Planets', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)
ax.legend(bbox_to_anchor=(1.01, 1.0), title='Planet Types', loc='upper left')
plt.show()

print(f'Total Raw Sample Size: {len(df)}')
print(f'Total Sample Size after filtering: {len(df_filtered)} ')
print(f'Total Sample Size after dropping nans: {len(df_stack)}')

from scipy import stats

chi2_stat, p_val_chi2, dof, expected_frequencies = stats.chi2_contingency(contingencytable)

print("--- Chi-Square Test of Independence ---")
print(f"Chi-Square Statistic (chi2): {chi2_stat:.3f}")
print(f"p-value: {p_val_chi2:.4e}")
print(f"Degrees of Freedom (dof): {dof}")

# Interpretation helper
if p_val_chi2 < 0.05:
    print("Result: Statistically significant association between variables (p < 0.05)")
else:
    print("Result: No statistically significant association; variables are independent (p >= 0.05)")
```

```python
!pip install adjustText

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from adjustText import adjust_text

# 1. Clean data locally for required parameters
dff = df_filtered.dropna(subset=['st_teff', 'pl_insol', 'pl_rade']).copy()

# 2. Fast vectorized classification using pd.cut
bins = [0, 1.0, 1.75, 3.8, 6.0, float('inf')]
labels = ['Terrestrial', 'Super Earth', 'Sub-Neptune', 'Neptunian', 'Jovian']
dff['Planet Classification'] = pd.cut(dff['pl_rade'], bins=bins, labels=labels)

# 3. Erratum Coefficients & Flux Calculation Function
coef_rv = [1.7763, 1.4335e-4, 3.3954e-9, -7.6364e-12, -1.1950e-15]
coef_rg = [1.0385, 1.2456e-4, 1.4612e-8, -7.6345e-12, -1.7511e-15]
coef_mg = [0.3507, 5.9578e-5, 1.6707e-9, -3.0058e-12, -5.1925e-16]
coef_em = [0.3207, 5.4471e-5, 1.5275e-9, -2.1709e-12, -3.8282e-16]

def calc_flux(coef, Ts):
    return coef[0] + coef[1]*Ts + coef[2]*Ts**2 + coef[3]*Ts**3 + coef[4]*Ts**4

# 4. Classify Habitability Regions
def determine_hz_region(row):
    Ts_star = row['st_teff'] - 5780
    p_flux_rv = calc_flux(coef_rv, Ts_star)
    p_flux_rg = calc_flux(coef_rg, Ts_star)
    p_flux_mg = calc_flux(coef_mg, Ts_star)
    p_flux_em = calc_flux(coef_em, Ts_star)

    flux = row['pl_insol']
    if flux > p_flux_rv:
        return 'Too Hot'
    elif p_flux_rv >= flux > p_flux_rg:
        return 'Optimistic HZ (Inner)'
    elif p_flux_rg >= flux >= p_flux_mg:
        return 'Conservative HZ'
    elif p_flux_mg > flux >= p_flux_em:
        return 'Optimistic HZ (Outer)'
    else:
        return 'Too Cold'

dff['HZ_Region'] = dff.apply(determine_hz_region, axis=1)

# 5. Export processed dataset to CSV
habitability_insol = dff[['pl_name', 'st_teff', 'pl_insol', 'Planet Classification', 'HZ_Region']]
habitability_insol.to_csv('Habitability Plot Data (Insolation).csv', index=False)

# 6. Initialize Figure
fig, ax = plt.subplots(figsize=(10, 6))

classify_color = {
    'Terrestrial': 'green',
    'Super Earth': 'red',
    'Sub-Neptune': 'lightblue',
    'Neptunian': 'darkblue',
    'Jovian': 'yellow'
}

classify_sizes = {
    'Terrestrial': 25,
    'Super Earth': 35,
    'Sub-Neptune': 45,
    'Neptunian': 70,
    'Jovian': 150
}

# Generate Habitability Zone Boundary Curves
T_star = np.linspace(2600, 7200, 100)
Ts = T_star - 5780

flux_rv = calc_flux(coef_rv, Ts)
flux_rg = calc_flux(coef_rg, Ts)
flux_mg = calc_flux(coef_mg, Ts)
flux_em = calc_flux(coef_em, Ts)

# Draw Background Zones
ax.fill_betweenx(T_star, 10.0, flux_rv, color='#ff4d4d', alpha=0.15, zorder=0, label='Too Hot')
ax.fill_betweenx(T_star, flux_rv, flux_em, color='#17b87e', alpha=0.2, label='Optimistic HZ', zorder=1)
ax.fill_betweenx(T_star, flux_rg, flux_mg, color='#00ff9d', alpha=0.3, label='Conservative HZ', zorder=2)
ax.fill_betweenx(T_star, flux_em, -1.0, color='#6688ff', alpha=0.15, zorder=0, label='Too Cold')

# Draw Limit Lines
ax.plot(flux_rv, T_star, color='#17b87e', linestyle='dashed', linewidth=1.5, alpha=0.8, zorder=3, label='Recent Venus Limit')
ax.plot(flux_rg, T_star, color='#00ff9d', linestyle='-', linewidth=2, zorder=3, label='Runaway Greenhouse Limit')
ax.plot(flux_mg, T_star, color='#00ff9d', linestyle=':', linewidth=2, zorder=3, label='Max Greenhouse Limit')
ax.plot(flux_em, T_star, color='#17b87e', linestyle='dashdot', linewidth=1.5, alpha=0.8, zorder=3, label='Early Mars Limit')

# Scatter Plot All Planets
for cat, color in classify_color.items():
    subset = dff[dff['Planet Classification'] == cat]
    ax.scatter(subset['pl_insol'], subset['st_teff'],
               c=color, s=classify_sizes[cat],
               alpha=0.6, edgecolor='black', label=cat, zorder=5)

# 7. Filter to label ONLY Terrestrial & Super-Earth planets inside HZ
target_classes = ['Terrestrial', 'Super Earth']
hz_regions = ['Optimistic HZ (Inner)', 'Conservative HZ', 'Optimistic HZ (Outer)']

hz_rocky_planets = dff[
    (dff['HZ_Region'].isin(hz_regions)) &
    (dff['Planet Classification'].isin(target_classes)) &
    (dff['pl_insol'].between(0.0, 2.2)) &
    (dff['st_teff'].between(2600, 7200))
]

texts = []
for _, planet in hz_rocky_planets.iterrows():
    cat = str(planet['Planet Classification'])
    texts.append(ax.text(
        planet['pl_insol'], planet['st_teff'], planet['pl_name'],
        fontsize=7, fontweight='bold',
        bbox=dict(boxstyle="round,pad=0.1", fc="white", ec=classify_color.get(cat, 'black'), alpha=0.85),
        zorder=7
    ))

# Set STANDARD non-inverted axes FIRST for adjustText calculation
ax.set_xlim(0.0, 2.2)
ax.set_ylim(2500, 7500)

# Adjust text without inverted coordinate feedback loop
if texts:
    adjust_text(
        texts,
        ax=ax,
        arrowprops=dict(arrowstyle="-", color='black', lw=0.5, alpha=0.7)
    )

# INVERT X-AXIS AFTER adjust_text finishes processing positions
ax.invert_xaxis()

# 8. Labels & Legend
ax.set_xlabel('Insolation Flux ($F_{\\oplus}$)', fontsize=12)
ax.set_ylabel('Host Star Effective Temperature (K)', fontsize=12)
ax.set_title('Habitability Zone of Solo-Exoplanetary Systems', fontsize=14, fontweight='bold')

ax.legend(loc='upper left', bbox_to_anchor=(1.02, 1), fontsize=10)
plt.tight_layout()
plt.show()
```

```python


import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from adjustText import adjust_text

# =========================================================
# 1. Start from the FINAL CORE FILTERED SAMPLE
# =========================================================

dff = df_filtered.dropna(
    subset=['st_teff', 'pl_insol', 'pl_rade']
).copy()

# Keep only the stellar-temperature range covered by the
# Kopparapu et al. (2013) HZ model
dff = dff[
    dff['st_teff'].between(2600, 7200)
].copy()


# =========================================================
# 2. Planet Classification
# =========================================================

bins = [
    0,
    1.0,
    1.7,
    3.5,
    6.0,
    float('inf')
]

labels = [
    'Terrestrial',
    'Super Earth',
    'Sub-Neptune',
    'Neptunian',
    'Jovian'
]

dff['Planet Classification'] = pd.cut(
    dff['pl_rade'],
    bins=bins,
    labels=labels
)


# =========================================================
# 3. Kopparapu et al. (2013) Erratum Coefficients
# =========================================================

coef_rv = [
    1.7763,
    1.4335e-4,
    3.3954e-9,
    -7.6364e-12,
    -1.1950e-15
]

coef_rg = [
    1.0385,
    1.2456e-4,
    1.4612e-8,
    -7.6345e-12,
    -1.7511e-15
]

coef_mg = [
    0.3507,
    5.9578e-5,
    1.6707e-9,
    -3.0058e-12,
    -5.1925e-16
]

coef_em = [
    0.3207,
    5.4471e-5,
    1.5275e-9,
    -2.1709e-12,
    -3.8282e-16
]


def calc_flux(coef, Ts):
    return (
        coef[0]
        + coef[1] * Ts
        + coef[2] * Ts**2
        + coef[3] * Ts**3
        + coef[4] * Ts**4
    )


# =========================================================
# 4. Determine HZ Region
# =========================================================

def determine_hz_region(row):

    Ts_star = row['st_teff'] - 5780

    p_flux_rv = calc_flux(coef_rv, Ts_star)
    p_flux_rg = calc_flux(coef_rg, Ts_star)
    p_flux_mg = calc_flux(coef_mg, Ts_star)
    p_flux_em = calc_flux(coef_em, Ts_star)

    flux = row['pl_insol']

    if flux > p_flux_rv:
        return 'Too Hot'

    elif p_flux_rv >= flux > p_flux_rg:
        return 'Optimistic HZ (Inner)'

    elif p_flux_rg >= flux >= p_flux_mg:
        return 'Conservative HZ'

    elif p_flux_mg > flux >= p_flux_em:
        return 'Optimistic HZ (Outer)'

    else:
        return 'Too Cold'


dff['HZ_Region'] = dff.apply(
    determine_hz_region,
    axis=1
)


# =========================================================
# 5. Keep only Terrestrial and Super-Earth planets
# =========================================================

target_classes = [
    'Terrestrial',
    'Super Earth'
]

hz_regions = [
    'Optimistic HZ (Inner)',
    'Conservative HZ',
    'Optimistic HZ (Outer)'
]

hz_rocky_planets = dff[
    (dff['HZ_Region'].isin(hz_regions)) &
    (dff['Planet Classification'].isin(target_classes))
].copy()


print(f'HZ plot sample: N = {len(dff)}')
print(
    f'Terrestrial/Super-Earth HZ candidates: '
    f'N = {len(hz_rocky_planets)}'
)

print(
    hz_rocky_planets[
        [
            'pl_name',
            'pl_rade',
            'st_teff',
            'pl_insol',
            'Planet Classification',
            'HZ_Region'
        ]
    ].sort_values('pl_insol')
)


# =========================================================
# 6. Export HZ candidate data
# =========================================================

habitability_insol = hz_rocky_planets[
    [
        'pl_name',
        'pl_rade',
        'st_teff',
        'pl_insol',
        'Planet Classification',
        'HZ_Region'
    ]
]

habitability_insol.to_csv(
    'Habitability_Plot_Data.csv',
    index=False
)


# =========================================================
# 7. Initialize Figure
# =========================================================

fig, ax = plt.subplots(
    figsize=(8, 6)
)

# Colors for planet classifications
classify_color = {
    'Terrestrial': 'green',
    'Super Earth': 'red'
}

classify_sizes = {
    'Terrestrial': 25,
    'Super Earth': 35
}


# =========================================================
# 8. Generate HZ Boundary Curves
# =========================================================

T_star = np.linspace(
    2600,
    7200,
    200
)

Ts = T_star - 5780

flux_rv = calc_flux(
    coef_rv,
    Ts
)

flux_rg = calc_flux(
    coef_rg,
    Ts
)

flux_mg = calc_flux(
    coef_mg,
    Ts
)

flux_em = calc_flux(
    coef_em,
    Ts
)


# =========================================================
# 9. Draw Background HZ Regions
# =========================================================

ax.fill_betweenx(
    T_star,
    10.0,
    flux_rv,
    color='#ff4d4d',
    alpha=0.15,
    zorder=0,
    label='Too Hot'
)

ax.fill_betweenx(
    T_star,
    flux_rv,
    flux_em,
    color='#17b87e',
    alpha=0.20,
    zorder=1,
    label='Optimistic HZ'
)

ax.fill_betweenx(
    T_star,
    flux_rg,
    flux_mg,
    color='#00ff9d',
    alpha=0.30,
    zorder=2,
    label='Conservative HZ'
)

ax.fill_betweenx(
    T_star,
    flux_em,
    -1.0,
    color='#6688ff',
    alpha=0.15,
    zorder=0,
    label='Too Cold'
)


# =========================================================
# 10. Draw HZ Boundary Lines
# =========================================================

ax.plot(
    flux_rv,
    T_star,
    color='#17b87e',
    linestyle='dashed',
    linewidth=1.5,
    alpha=0.8,
    zorder=3,
    label='Recent Venus Limit'
)

ax.plot(
    flux_rg,
    T_star,
    color='#00ff9d',
    linestyle='-',
    linewidth=2,
    zorder=3,
    label='Runaway Greenhouse Limit'
)

ax.plot(
    flux_mg,
    T_star,
    color='#00ff9d',
    linestyle=':',
    linewidth=2,
    zorder=3,
    label='Maximum Greenhouse Limit'
)

ax.plot(
    flux_em,
    T_star,
    color='#17b87e',
    linestyle='dashdot',
    linewidth=1.5,
    alpha=0.8,
    zorder=3,
    label='Early Mars Limit'
)


# =========================================================
# 11. Plot ONLY Terrestrial + Super-Earth planets
# =========================================================

for cat, color in classify_color.items():

    subset = dff[
        dff['Planet Classification'] == cat
    ]

    ax.scatter(
        subset['pl_insol'],
        subset['st_teff'],
        color=color,
        s=classify_sizes[cat],
        alpha=0.6,
        edgecolor='black',
        label=cat,
        zorder=5
    )


# =========================================================
# 12. Label ONLY HZ candidates
# =========================================================

texts = []

for _, planet in hz_rocky_planets.iterrows():

    cat = str(
        planet['Planet Classification']
    )

    texts.append(
        ax.text(
            planet['pl_insol'],
            planet['st_teff'],
            planet['pl_name'],
            fontsize=7,
            fontweight='bold',
            bbox=dict(
                boxstyle="round,pad=0.1",
                fc="white",
                ec=classify_color.get(cat, 'black'),
                alpha=0.85
            ),
            zorder=7
        )
    )


# =========================================================
# 13. Axis Configuration
# =========================================================

ax.set_xlim(
    0.0,
    2.2
)

ax.set_ylim(
    2600,
    7200
)

# Run adjustText BEFORE inverting x-axis
if texts:

    adjust_text(
        texts,
        ax=ax,
        arrowprops=dict(
            arrowstyle="-",
            color='black',
            lw=0.5,
            alpha=0.7
        )
    )

ax.invert_xaxis()


# =========================================================
# 14. Labels and Legend
# =========================================================

ax.set_xlabel(
    'Insolation Flux ($F_{\\oplus}$)',
    fontsize=12
)

ax.set_ylabel(
    'Host Star Effective Temperature (K)',
    fontsize=12
)

ax.set_title(
    'Terrestrial and Super-Earth Planets Within the Conventional Habitable Zone',
    fontsize=14,
    fontweight='bold'
)

ax.legend(
    loc='upper left',
    bbox_to_anchor=(1.02, 1),
    fontsize=9
)

ax.grid(
    True,
    linestyle='--',
    alpha=0.5
)

plt.tight_layout()
plt.show()
```

```python
from scipy.stats import kruskal, mannwhitneyu
from itertools import combinations
from statsmodels.stats.multitest import multipletests


stats_summary = (
    df_viol
    .groupby('Planet Classification', observed=True)['st_met']
    .describe()
)

print('Descriptive Statistics:')
print(stats_summary)


#KRUSKAL-WALLIS TEST


groups = [
    group['st_met'].values
    for name, group in df_viol.groupby(
        'Planet Classification',
        observed=True
    )
]

H, p = kruskal(*groups)

k = len(groups)
n = len(df_viol)

epsilon_squared = (H - k + 1) / (n - k)

print('\nKruskal-Wallis Test:')
print(f'H-statistic: {H:.4f}')
print(f'p-value: {p:.6e}')
print(f'Epsilon-squared: {epsilon_squared:.4f}')



# PAIRWISE MANN-WHITNEY U TESTS


results = []

group_names = [
    group for group in labels
    if group in df_viol['Planet Classification'].unique()
]

for group1, group2 in combinations(group_names, 2):

    values1 = df_viol[
        df_viol['Planet Classification'] == group1
    ]['st_met']

    values2 = df_viol[
        df_viol['Planet Classification'] == group2
    ]['st_met']

    U, p_pair = mannwhitneyu(
        values1,
        values2,
        alternative='two-sided'
    )

    results.append({
        'Group 1': group1,
        'Group 2': group2,
        'U statistic': U,
        'Raw p-value': p_pair
    })


pairwise_results = pd.DataFrame(results)



# HOLM MULTIPLE-COMPARISON CORRECTION


reject, p_corrected, _, _ = multipletests(
    pairwise_results['Raw p-value'],
    method='holm'
)

pairwise_results['Holm p-value'] = p_corrected
pairwise_results['Significant'] = reject

print('\nPairwise Mann-Whitney U Tests:')
print(pairwise_results)
```
